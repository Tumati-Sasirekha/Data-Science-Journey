# Data Types

Exploring Python's built-in data types: numbers, text, booleans, collections, and type conversion.

## 1. Numeric types (int, float, complex)

```python
a = 42
b = 3.14
c = 2 + 3j

print(a, type(a))
print(b, type(b))
print(c, type(c))
```

## 2. Text type (str)

```python
s = "Data Science"
print(s, type(s))
print(s[0])       # indexing
print(s[0:4])     # slicing
print(s.upper())
print(len(s))
```

## 3. Boolean type

```python
print(True, type(True))
print(bool(0), bool(1), bool(""), bool("hi"), bool([]), bool([1]))  # truthy/falsy values
```

## 4. NoneType

```python
x = None
print(x, type(x))
print(x is None)   # correct way to check for None
```

## 5. Lists (ordered, mutable)

```python
fruits = ["apple", "banana", "cherry"]
print(fruits, type(fruits))

fruits.append("date")
fruits[0] = "avocado"
print(fruits)
```

## 6. Tuples (ordered, immutable)

```python
point = (10, 20)
print(point, type(point))

try:
    point[0] = 99
except TypeError as e:
    print("Error:", e)
```

## 7. Dictionaries (key-value pairs)

```python
student = {"name": "Sam", "age": 21, "major": "Statistics"}
print(student, type(student))

print(student["name"])
print(student.keys())
print(student.values())
print(student.items())
```

## 8. Sets (unique, unordered)

```python
nums = {1, 2, 2, 3, 3, 3}
print(nums, type(nums))   # duplicates dropped automatically

a = {1, 2, 3}
b = {2, 3, 4}
print(a | b)   # union
print(a & b)   # intersection
print(a - b)   # difference
```

## 9. Checking types: type() vs isinstance()

```python
x = 10

print(type(x) == int)
print(isinstance(x, int))
print(isinstance(x, (int, float)))   # isinstance can check multiple types at once
```

## 10. Type casting

```python
print(int("5") + 10)      # str -> int
print(str(5) + " apples")  # int -> str
print(float("3.14") * 2)   # str -> float
print(list("abc"))         # str -> list of characters
print(tuple([1, 2, 3]))    # list -> tuple
```
