# Reading User Input

Exploring input() and how to read/convert it into different data types.

## 1. input() always returns a string

```python
value = input("Enter something: ")
print(value, type(value))
```

## 2. Converting input to int

```python
age = int(input("Enter your age: "))
print(age, type(age))
print(age + 5)   # works because it's now a real int
```

## 3. Converting input to float

```python
price = float(input("Enter a price: "))
print(price, type(price))
print(price * 1.18)   # e.g. add 18% tax
```

## 4. Reading two values on one line (manual unpacking)

```python
a, b = input("Enter two numbers separated by a space: ").split()
a, b = int(a), int(b)
print(a, b, a + b)
```

## 5. Same thing using map()

```python
a, b = map(int, input("Enter two numbers: ").split())
print(a, b, a + b)
```

## 6. Reading a list of numbers


```python
nums = list(map(int, input("Enter several numbers separated by spaces: ").split()))
print(nums, type(nums))
print(sum(nums))
```

## 7. Handling invalid input with try/except

```python
try:
    age = int(input("Enter your age: "))
    print("Age:", age)
except ValueError:
    print("That's not a valid number — please enter digits only.")
```

## 8. Why bool(input(...)) doesn't do what you'd expect

```python
answer = input("Type 'False' and see what happens: ")
print(bool(answer))   # True! Any non-empty string is truthy

# correct way to check yes/no style input
is_yes = answer.strip().lower() == "yes"
print("is_yes:", is_yes)
```

## Summary
- input() → always str
- cast explicitly: int(), float()
- split() + map() for multiple values
- try/except for invalid input
- compare strings manually for boolean-style input
