# Operators — Complete Guide

From arithmetic basics through bitwise, walrus, and chained comparisons.

## 1. Arithmetic operators

```python
a, b = 5, 2
print(a + b)
print(a - b)
print(a * b)
print(a / b)   # true division
print(a // b)  # floor division
print(a % b)   # modulus
print(a ** b)  # exponent
```

## 2. Comparison operators

```python
print(5 == 5)
print(5 != 2)
print(5 > 2)
print(5 < 2)
print(5 >= 5)
print(5 <= 4)
```

## 3. Logical operators

```python
x, y = True, False
print(x and y)
print(x or y)
print(not x)
```

## 4. Assignment operators

```python
count = 10
count += 5
print(count)
count -= 3
print(count)
count *= 2
print(count)
count //= 4
print(count)
```

## 5. Membership operators

```python
print("a" in "apple")
print(5 not in [1, 2, 3])
print("key" in {"key": "value"})   # checks keys, not values, for dicts
```

## 6. Identity vs equality

```python
list1 = [1, 2, 3]
list2 = [1, 2, 3]
print(list1 == list2)  # True, same values
print(list1 is list2)  # False, different objects in memory

list3 = list1
print(list1 is list3)  # True, same object
```

## 7. Bitwise operators

```python
a, b = 5, 3   # binary: 101, 011
print(a & b)   # AND  -> 001 -> 1
print(a | b)   # OR   -> 111 -> 7
print(a ^ b)   # XOR  -> 110 -> 6
print(~a)      # NOT  -> -6
print(a << 1)  # left shift -> 10
print(a >> 1)  # right shift -> 2
```

## 8. Operator precedence

```python
result = 2 + 3 * 4       # multiplication happens first
print(result)             # 14

result2 = (2 + 3) * 4     # parentheses override precedence
print(result2)             # 20

result3 = 2 ** 3 ** 2     # exponent is right-associative: 2 ** (3 ** 2)
print(result3)             # 512
```

## 9. Chained comparisons

```python
x = 5
print(1 < x < 10)          # True
print(10 < x < 20)         # False

grade = 85
print("B" if 70 <= grade < 90 else "not B")
```

## 10. Ternary (conditional) operator

```python
age = 20
status = "adult" if age >= 18 else "minor"
print(status)

# nested ternary (use sparingly - readability drops fast)
score = 75
grade = "A" if score >= 90 else "B" if score >= 75 else "C"
print(grade)
```

## 11. Walrus operator (:=)

```python
data = [1, 2, 3, 4, 5]

# without walrus - len() computed, then used
n = len(data)
if n > 3:
    print(f"List has {n} items")

# with walrus - assign and check in one line
if (n := len(data)) > 3:
    print(f"List has {n} items, which is more than 3")
```

## Summary
- Arithmetic, comparison, logical, assignment, membership, identity — the everyday six
- Bitwise, precedence rules, chained comparisons, ternary, and walrus — the advanced set
- Operator overloading (customizing `+`, `==` for your own classes) comes later with OOP
