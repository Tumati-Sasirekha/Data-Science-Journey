# Variables

Exploring how variables work in Python: assignment, reassignment, multiple assignment, and related concepts.

## 1. Basic assignment

```python
name = "Alice"
age = 25
height = 5.6

print(name, age, height)
```

## 2. Dynamic typing
A variable's type is inferred from its value and can change on reassignment.

```python
x = 10
print(x, type(x))

x = "now I'm a string"
print(x, type(x))
```

## 3. Multiple assignment
Assign several variables in a single line.

```python
a, b, c = 1, 2, 3
print(a, b, c)
```

## 4. Swapping variables
Python lets you swap values without a temporary variable.

```python
a, b = 1, 2
print("before:", a, b)

a, b = b, a
print("after:", a, b)
```

## 5. Chained assignment
Assign the same value to multiple variables at once.

```python
x = y = z = 100
print(x, y, z)

y = 50
print(x, y, z)  # only y changes — they aren't linked after assignment
```

## 6. Augmented assignment operators
Shorthand for updating a variable based on its current value.

```python
count = 10
count += 5   # count = count + 5
print(count)

count -= 3
print(count)

count *= 2
print(count)

count /= 4
print(count)
```

## 7. Naming conventions and constants
Variables use snake_case. Constants (values not meant to change) use UPPER_CASE by convention — Python doesn't enforce this, it's just a signal to other readers.

```python
student_name = "Sam"       # normal variable, snake_case
PI = 3.14159                # constant by convention
MAX_ATTEMPTS = 3

print(student_name, PI, MAX_ATTEMPTS)
```

## 8. Checking identity with id()
`id()` shows the memory address of an object. Two variables with the same simple value may point to the same object.

```python
a = 10
b = 10
print(id(a), id(b), a is b)   # small ints are often cached, so this is often True

c = [1, 2, 3]
d = [1, 2, 3]
print(id(c), id(d), c is d)   # separate list objects, so this is False
```

## 9. Deleting variables
`del` removes a variable. Using it afterward raises a NameError.

```python
temp = "I'll be deleted"
print(temp)

del temp

try:
    print(temp)
except NameError as e:
    print("Error:", e)
```

## 10. f-strings for readable printing
f-strings let you embed variables directly inside a string.

```python
name = "Alice"
age = 25

print(f"{name} is {age} years old")
print(f"Next year, {name} will be {age + 1}")
```
